2. Data Visualization

 Graph 1: Survival by gender

```python
# Passenger survival
```

```python
grouped = train.groupby('Sex')['Survived'].agg(['mean', 'sum', 'count'])
grouped
```

```python
plt.figure(figsize=(5,3))
sns.barplot(data=train, x='Sex', y='Survived', palette='Set2')
plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Survived')
plt.legend(['Not Servived', 'Sevived'])
plt.show()
```

Gender: Female survival rate-74% and Male-19%

replace null value with N

```python
train.fillna({'Embarked': "N"})
```

Graph 2: Survival by Class

```python
plt.figure(figsize=(6,5))
sns.barplot(data=train, x='Sex',y='Pclass', hue='Survived', palette='Set2')
plt.title('Survival Rate by  Passenger Class ')
plt.xlabel('Gender')
plt.ylabel('class')
plt.legend(['Sevived Rate'])
plt.show()
```

Class : 1st class survived greater than others

Graph 3: Age Distribution

```python
train_filtered = train[train['Age']>30]
print(len(train_filtered))
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data=train, x='Age',hue='Survived', multiple='stack', bins=30, color='blue')
plt.title('Age Distribution:Survived vs Non-Servived')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

Graph 4: Heatmap

```python
plt.figure(figsize=(6,5))
sns.heatmap(train.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

This correlation heatmap shows how Titanic  dataset survival has the strongest nagetive correlation with Pclass-0.34 and possitive correlation with fare 0.26.

Key insight - CrossTab

```python
pd.crosstab(train['Pclass'], train['Survived'])
plt.plot(kind='bar', stacked=True)
plt.title('Survival count by class')
plt.show()
```
